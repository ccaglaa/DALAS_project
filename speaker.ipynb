```python
import pandas as pd
import glob

df = pd.read_csv('only_talks.csv')
print(f"Dataset shape: {df.shape}")
print(f"Columns: {df.columns.tolist()}")
df.head()
#df_test = df.head(10)
```

```python
import pandas as pd
import re


def clean_and_split_speakers(speaker_string):
    """
    Nettoie et sépare les noms de speakers
    """
    if pd.isna(speaker_string):
        return []
    
    # Standardiser les séparateurs
    speaker_string = str(speaker_string)
    speaker_string = re.sub(r'\s*&\s*|\s+and\s+|\s+et\s+|\s+und\s+|\s+y\s+', ', ', speaker_string, flags=re.IGNORECASE)
    
    # Supprimer les espaces multiples
    speaker_string = re.sub(r'\s+', ' ', speaker_string)
    
    # Séparer par virgule
    speakers = [s.strip() for s in speaker_string.split(',') if s.strip()]
    
    return speakers

# Ajouter la liste des speakers à la main df pour chaque video (on pourrait remplacer la colonne speaker par ça)
df['speakers_list'] = df['speaker'].apply(clean_and_split_speakers)

# Vérifier les résultats
df.head()
```

```python
# Créer une liste pour le nouveau DataFrame
expanded_rows = []

for idx, row in df.iterrows():
    video_id = row['id']
    views = row['views']
    
    for speaker_name in row['speakers_list']:
        expanded_rows.append({
            'speaker': speaker_name,
            'video_id': video_id,
            'views': views,
            'topics': row['topics'],
        })

# Créer le DataFrame étendu
expanded_df = pd.DataFrame(expanded_rows)
print(f"Nombre total d'entrées : {len(expanded_df)}")
print(f"Nombre de speakers uniques : {expanded_df['speaker'].nunique()}")

expanded_df.head()
```

```python
def aggregate_speaker_data(expanded_df):
    """
    Agrége les données par speaker pour obtenir le format demandé
    """
    # Fonction pour combiner les listes de topics uniques
    def combine_topics(series):
        # Si les topics sont déjà sous forme de liste
        all_topics = []
        for item in series:
            if isinstance(item, list):
                all_topics.extend(item)
            elif pd.notna(item):
                # Si c'est une chaîne, on peut la traiter
                all_topics.append(str(item))
        # Retirer les doublons
        return list(set(all_topics))
    
    # Agréger les données
    speaker_agg = expanded_df.groupby('speaker').agg(
        video_ids=('video_id', list),  # Liste des IDs de vidéos
        videos=('video_id', 'count'),  # Nombre de vidéos
        views=('views', 'sum'),  # Somme des vues
        topics=('topics', combine_topics)  # Tous les topics uniques
    ).reset_index()
    
    # Calculer la moyenne des vues
    speaker_agg['average_views'] = speaker_agg['views'] / speaker_agg['videos']
    
    # Arrondir la moyenne
    speaker_agg['average_views'] = speaker_agg['average_views'].round(2)
    
    # Trier par nombre de vidéos (descendant)
    speaker_agg = speaker_agg.sort_values('videos', ascending=False)
    
    return speaker_agg

# Créer le DataFrame final des speakers
speaker_df = aggregate_speaker_data(expanded_df)



speaker_df.columns
speaker_df.shape

speaker_df.t_csv("speaker_dff.csv")
```
